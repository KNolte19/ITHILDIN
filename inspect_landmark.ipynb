```python
df = pd.read_csv("training/data_psycho/landmark/landmark_annotations_benin.csv", sep=";")
filename = "psychodidae_benin-sandfly-sampling_m_l_1b6c687c4619.tif"
image = plt.imread(os.path.join("training/data_psycho/landmark/image_raw", filename))
df_sample = df[df["filename"] == filename]

x_ls, y_ls = [], []
for i in range(0, 12):
    print(i, df_sample[f"X_{i}"].values[0])
    x_ls.append(df_sample[f"X_{i}"].values[0])
    y_ls.append(df_sample[f"Y_{i}"].values[0])

plt.imshow(image)
plt.scatter(x_ls, y_ls, c="red", marker="o", s=3)

# Annotate the points with their index
for i in range(0, 12):
    plt.text(x_ls[i], y_ls[i], str(i), color="black", fontsize=12)

#plt.axis("off")
plt.tight_layout()
```
